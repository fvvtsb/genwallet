# Săn Ví BEP20 Đuôi 88888888 An Toàn Bằng Public Key (Trustless)
Quy trình này đảm bảo an toàn tuyệt đối 100%. Bạn tự sinh Private Key ở máy cá nhân (offline), sau đó chỉ cung cấp **Public Key (128 ký tự hex)** lên đây cho GPU chạy.

```python
# 1. Tải thư viện & Biên dịch Profanity2
!nvidia-smi
!apt-get update && apt-get install -y ocl-icd-opencl-dev gcc g++ make
!rm -rf profanity2 && git clone https://github.com/1inch/profanity2.git
!cd profanity2 && make
```

```python
# 2. NHẬP SEED PUBLIC KEY CỦA BẠN
import os
pub_key = input("Hãy nhập Seed Public Key của bạn (128 ký tự Hex, không có tiền tố 0x04): ").strip()
if len(pub_key) != 128:
    print("⚠️ CẢNH BÁO: Public Key tiêu chuẩn phải dài đúng 128 ký tự Hex!")
else:
    print("✅ Đã nhận Public Key hợp lệ.")
os.environ['PUB_KEY'] = pub_key
```

```python
# 3. KÍCH HOẠT GPU TÌM KIẾM
# Nó sẽ khớp với 8 số 8 ở cuối (88888888)
!cd profanity2 && ./profanity2.x64 --matching XXXXXXXXXXXXXXXXXXXXXXXXXXXXXXXX88888888 -z $PUB_KEY
```

## Bước cuối cùng (Thực hiện offline trên PC của bạn)
Khi ô trên chạy xong và in ra `Private key: <Modifier>`, bạn hãy mang đoạn chữ số đó về máy tính của mình và làm phép toán Elip:
**`Private Key Của Ví = (Seed Private Key Của Bạn + Modifier) % N`**

*(Tuyệt đối không dán Private Key của bạn lên đây để tính toán)*
